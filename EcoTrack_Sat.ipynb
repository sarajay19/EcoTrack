```python
# Import the libraries
from glob import glob
import earthpy as et
import earthpy.spatial as es
import earthpy.plot as ep
import rasterio as rio
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import ListedColormap
import plotly.graph_objects as go
```

```python
# Reading the data
np.seterr(divide='ignore', invalid='ignore')
S_sentinel_bands = glob("/content/satellite-imagery.png.webp")
S_sentinel_bands.sort()
```

```python
# Visualize Bands
l = []
for i in S_sentinel_bands:
    with rio.open(i, 'r') as f:
        l.append(f.read(1))
arr_st = np.stack(l)
ep.plot_bands(arr_st, cmap='gist_earth', figsize=(20, 12), cols=6, cbar=False)
plt.show()
```

```python
# RGB Composite Image
print(arr_st.shape)  # Check the shape of the array
```

```python
# Single Band Visualization
ep.plot_bands(arr_st, cmap='gist_earth', figsize=(10, 10), cbar=True)
plt.title("Single Band Visualization")
plt.show()
```

```python
# Create a dummy RGB by replicating the single band
rgb_dummy = np.repeat(arr_st, 3, axis=0)  # Repeat the single band 3 times
```

```python
# Plot the dummy RGB image
ep.plot_rgb(rgb_dummy, rgb=(0, 1, 2), figsize=(10, 16))
plt.title("Dummy RGB Composite from Single Band")
plt.show()
```

```python
# Single Band Visualized as RGB
ep.plot_rgb(arr_st,
            rgb=(0, 0, 0),  # Use the single band for R, G, and B
            stretch=True,
            str_clip=0.2,
            figsize=(10, 16))
plt.title("Single Band Visualized as RGB")
plt.show()
```

```python
# Histogram of Bands
colors = ['tomato', 'navy', 'MediumSpringGreen', 'lightblue', 'orange', 'blue',
          'maroon', 'purple', 'yellow', 'olive', 'brown', 'cyan']

# Provide a single title for the histogram
ep.hist(arr_st,
         colors=colors,
         title='Histogram of Bands',  # Single title
         cols=3,
         alpha=0.5,
         figsize=(12, 10))
plt.show()
```

```python
# Visualize the single band
ep.plot_bands(arr_st, cmap='gist_earth', figsize=(10, 10), cbar=True)
plt.title("Single Band Visualization")
plt.show()
```

```python
# Create dummy bands by repeating the single band
dummy_band_red = arr_st[0]  # Using the single band as the red channel
dummy_band_nir = arr_st[0]  # Using the same band as the NIR channel
dummy_band_green = arr_st[0]  # Using the same band as the green channel

# Example: NDVI Calculation with dummy bands
ndvi = (dummy_band_nir - dummy_band_red) / (dummy_band_nir + dummy_band_red + 1e-10)  # Avoid division by zero
ep.plot_bands(ndvi, cmap="RdYlGn", cols=1, vmin=-1, vmax=1, figsize=(10, 14))
plt.title("Dummy NDVI")
plt.show()
```

```python
# Visualize the single band
ep.plot_bands(arr_st, cmap='gist_earth', figsize=(10, 10), cbar=True)
plt.title("Single Band Visualization")
plt.show()
```

```python


# MNDWI Calculation
mndwi = es.normalized_diff(arr_st[2], arr_st[10])
ep.plot_bands(mndwi, cmap="RdYlGn", cols=1, vmin=-1, vmax=1, figsize=(10, 14))
plt.show()

# NDMI Calculation
ndmi = es.normalized_diff(arr_st[7], arr_st[10])
ep.plot_bands(ndmi, cmap="RdYlGn", cols=1, vmin=-1, vmax=1, figsize=(10, 14))
plt.show()

# CMR Calculation
cmr = np.divide(arr_st[10], arr_st[11])
ep.plot_bands(cmr, cmap="RdYlGn", cols=1, vmin=-1, vmax=1, figsize=(10, 14))
plt.show()

# FMR Calculation
fmr = np.divide(arr_st[10], arr_st[7])
ep.plot_bands(fmr, cmap="RdYlGn", cols=1, vmin=-1, vmax=1, figsize=(10, 14))
plt.show()
```
